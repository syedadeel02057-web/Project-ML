```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
X = df.drop(columns='Attrition')
y = df['Attrition'].map({'No': 0, 'Yes': 1})
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
x_train,x_test,y_train,y_test = train_test_split(X, y, test_size=0.30, random_state=42,stratify = y)
```

```python
df.sample()
```

```python
nominal_cols = [
    'BusinessTravel',
    'Department',
    'EducationField',
    'Gender',
    'JobRole',
    'MaritalStatus',
    'OverTime'
]
num_cols = [
    'Age',
    'DailyRate',
    'DistanceFromHome',
    'HourlyRate',
    'MonthlyIncome',
    'MonthlyRate',
    'NumCompaniesWorked',
    'PercentSalaryHike',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]
ordinal_cols = [
    'BusinessTravel'
]
```

```python
num_feature = df.select_dtypes(include = 'number')
Q1 = num_feature.quantile(0.25)
Q3 = num_feature.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = (num_feature < lower_bound) | (num_feature > upper_bound)
print(outliers.sum())
```

```python
nominal_cols_pipeline = Pipeline(steps = [
             ('nominal_encode',OneHotEncoder(categories = 'auto',handle_unknown= 'ignore'))
])
ordinal_cols_pipeline = Pipeline(steps = [
    ('ordinal_encode',OrdinalEncoder(categories = 'auto'))
])
num_cols_pipeline = Pipeline(steps = [
    ('scale',StandardScaler())
])


```

```python
preprocessor = ColumnTransformer(transformers=[
    ("nominal_encode", nominal_cols_pipeline, nominal_cols),
    ("ordinal_encode", ordinal_cols_pipeline, ordinal_cols),
    ("scale", num_cols_pipeline, num_cols),
    (remainder='passthrough'),
])
full_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42))
])
```

```python
full_pipeline.fit(x_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.ensemble import RandomForestClassifier
```

```python
y_pred = full_pipeline.predict(x_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
full_pipeline_rf = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(class_weight='balanced', random_state=42))
])
full_pipeline_rf.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from sklearn.model_selection import cross_val_score

rf_cv_scores = cross_val_score(full_pipeline_rf, x_train, y_train, cv=cv, scoring='accuracy')

print("Fold accuracies:", rf_cv_scores)
print("Mean accuracy:", rf_cv_scores.mean())
```
